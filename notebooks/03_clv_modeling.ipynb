# CLV Modelling

Train a baseline regression model and inspect MAE, RMSE, and R².

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from src.customer360.features import build_features
from src.customer360.models import train_clv_model, regression_metrics

df = pd.read_csv('../data/sample/customers.csv')
X = build_features(df)
X_train, X_test, y_train, y_test = train_test_split(X, df.future_value, test_size=0.2, random_state=42)
model = train_clv_model(X_train, y_train)
pred = model.predict(X_test)
regression_metrics(y_test, pred)
```
