# Churn Modelling

Estimate churn probabilities and evaluate classification performance.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from src.customer360.features import build_features
from src.customer360.models import train_churn_model, classification_metrics

df = pd.read_csv('../data/sample/customers.csv')
X = build_features(df)
X_train, X_test, y_train, y_test = train_test_split(X, df.churned, test_size=0.2, random_state=42, stratify=df.churned)
model = train_churn_model(X_train, y_train)
prob = model.predict_proba(X_test)[:,1]
pred = (prob >= 0.5).astype(int)
classification_metrics(y_test, prob, pred)
```
